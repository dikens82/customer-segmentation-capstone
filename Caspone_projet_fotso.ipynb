**3.4 Feature Engineering using RFM Analysis**

RFM analysis was used to transform transactional data into customer-level behavioral features.

**Define Snapshot Date**

```
snapshot_date = df['ORDERDATE'].max() + pd.Timedelta(days=1)

print(snapshot_date)
```

```
rfm = df.groupby('CUSTOMERNAME').agg({
    'ORDERDATE': lambda x: (snapshot_date - x.max()).days,
    'ORDERNUMBER': 'nunique',
    'SALES': 'sum'
})
```

```
rfm.columns = ['Recency','Frequency','Monetary']

rfm.head()
```

Step 4: Modeling, Evaluation and Business Interpretation

4.1 Clustering Algorithms Evaluated

Three clustering algorithms were applied and evaluated on the RFM dataset:

K-Means Clustering
Agglomerative Hierarchical Clustering
DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

Since this is an unsupervised learning problem, the Silhouette Score was selected as the primary evaluation metric. The Silhouette Score measures both cluster cohesion and cluster separation. A higher score indicates better-defined clusters.



**4.2 K-Means Clustering**

**Determining the Optimal Number of Clusters**

The Elbow Method was used to identify the optimal value of K.

```
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2,11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(rfm_pca)

    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8,5))

plt.plot(range(2,11), inertia, marker='o')

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.title("Elbow Method")

plt.show()
```

**Interpretation**

The elbow point appears at K = 4, indicating that four clusters provide a suitable balance between simplicity and cluster compactness.



**Training K-Means**

```
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(rfm_pca)
```

**Silhouette Score**

```
from sklearn.metrics import silhouette_score

kmeans_score = silhouette_score(
    rfm_pca,
    kmeans_labels
)

print("K-Means Silhouette Score:", kmeans_score)
```



**4.3 Hierarchical (Agglomerative) Clustering**


Dendrogram Visualization

```
from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(
    rfm_pca,
    method='ward'
)

plt.figure(figsize=(12,6))

dendrogram(linked)

plt.title("Hierarchical Clustering Dendrogram")

plt.show()
```

**Interpretation**

The dendrogram suggests a four-cluster solution similar to K-Means.

**Model Training**

```
from sklearn.cluster import AgglomerativeClustering

agglomerative = AgglomerativeClustering(
    n_clusters=4
)

agg_labels = agglomerative.fit_predict(rfm_pca)
```

**Silhouette Score**

```
agg_score = silhouette_score(
    rfm_pca,
    agg_labels
)

print(
    "Agglomerative Silhouette Score:",
    agg_score
)
```

**4.4 DBSCAN Clustering**

**Model Training **

```
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.5,
    min_samples=5
)

dbscan_labels = dbscan.fit_predict(rfm_pca)
```

**Silhouette Score**

```
if len(set(dbscan_labels)) > 1:

    dbscan_score = silhouette_score(
        rfm_pca,
        dbscan_labels
    )

    print(
        "DBSCAN Silhouette Score:",
        dbscan_score
    )
```

**4.5 Model Comparison**

```
comparison = pd.DataFrame({
    "Algorithm":[
        "K-Means",
        "Agglomerative",
        "DBSCAN"
    ],
    "Silhouette Score":[
        kmeans_score,
        agg_score,
        dbscan_score
    ]
})

comparison.sort_values(
    by="Silhouette Score",
    ascending=False
)
```

Interpretation

DBSCAN achieved the highest Silhouette Score (0.565), followed by K-Means (0.528) and Agglomerative Clustering (0.508). Since a higher Silhouette Score indicates better cluster separation and cohesion, DBSCAN was selected as the best-performing model.

The results suggest that DBSCAN produced the most distinct customer groups within the dataset.

**STEP 5: CRITICAL THINKING, ETHICAL AI, AND BIAS AUDITING**

**5.1 Ethical AI Considerations**

Customer segmentation models can unintentionally introduce biases that disproportionately affect certain groups of customers. Therefore, an ethical AI assessment was conducted to evaluate whether the clustering process produced unequal treatment across different customer populations.

Traditional fairness metrics commonly used in supervised machine learning were considered.

These metrics include:

Demographic Parity
Equal Opportunity
Equalized Odds
Disparate Impact

However, these metrics cannot be applied to this dataset because the Sales Data Sample dataset does not contain demographic attributes such as:

Gender
Race
Ethnicity
Age
Income

Since no protected attributes are available, standard fairness metrics cannot be computed.

Therefore, proxy variables were used to investigate potential clustering bias.

The two most relevant variables available in the dataset are:

Country
Territory

These variables allow an assessment of whether specific geographic groups are disproportionately assigned to particular customer segments.

**5.2 Bias Audit Using Country**
